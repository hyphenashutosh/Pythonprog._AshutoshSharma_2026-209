```python
#break-stop a loop early
#break immediatley stops a whole loop even if there are items left


for i in range (1,10):
  if i==5:
    break
  print(i)
```

print 1 to 5 but skip 3

```python
for i in range(1,6):
  if i==3:
    continue
  print(i)
```
